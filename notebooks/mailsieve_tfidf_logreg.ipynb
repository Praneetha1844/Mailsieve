# Mailsieve: spam vs. ham email classifier

TF-IDF features with a logistic regression classifier. Label `1` is spam and `0` is ham in `data/spam_or_not_spam.csv`.

```python
import numpy as np                                                        #importing all the necessary modules for the project
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
```

```python
df=pd.read_csv("../data/spam_or_not_spam.csv")                  #Reading the sample data
```

```python
print(df)                                                                             #glimpse of sample data
```

```python
data=df.where(pd.notnull(df),'')                                           #checking removing null values
```

```python
data.info()                                                                #checking basic information about the data
```

```python
x= data['email']
y=data['label']                                                                
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(x,y,test_size=0.2,random_state=3)              #splittong the data into train and test
```

```python
print(x.shape)
print(X_train.shape)                                                       #nalysing the shape of x and y
print(X_test.shape)
```

```python
print(y.shape)
print(Y_train.shape)
print(Y_test.shape)
```

```python
feature_extraction=TfidfVectorizer(min_df=1,stop_words='english',lowercase=True)

X_train_features=feature_extraction.fit_transform(X_train)
X_test_features=feature_extraction.transform(X_test)

Y_train=Y_train.astype('int')
Y_test=Y_test.astype('int')
```

```python
print(X_train_features.shape)                                                #documents x vocabulary size
```

```python
model=LogisticRegression()
```

```python
model.fit(X_train_features,Y_train)
```

```python
prediction_on_training_data=model.predict(X_train_features)
accuracy_on_training_data=accuracy_score(Y_train,prediction_on_training_data)
```

```python
print('accuracy score on train data is:',accuracy_on_training_data)
```

```python
prediction_on_test_data =model.predict(X_test_features)
accuracy_on_test_data=accuracy_score(Y_test,prediction_on_test_data)
```

```python
print('accuracy score on test data is',accuracy_on_test_data)
```

## Beyond accuracy

The classes are imbalanced (about 83% ham), so accuracy alone can look good while spam is missed. These cells report precision, recall and F1 per class and the confusion matrix for the model trained above. They have not been executed in this saved notebook.

```python
from sklearn.metrics import classification_report, confusion_matrix

print(classification_report(Y_test, prediction_on_test_data, target_names=['ham', 'spam']))
print('rows = true label, columns = predicted label (ham, spam)')
print(confusion_matrix(Y_test, prediction_on_test_data))
```

## Re-check on de-duplicated data

The CSV contains exact duplicate emails, so a test email can also appear in the training split. This cell removes duplicates before splitting, stratifies on the label and re-trains, to show how much the score depends on that overlap. It has not been executed in this saved notebook.

```python
unique = data.drop_duplicates(subset='email')
x_unique = unique['email']
y_unique = unique['label'].astype('int')

Xu_train, Xu_test, Yu_train, Yu_test = train_test_split(
    x_unique, y_unique, test_size=0.2, random_state=3, stratify=y_unique
)

vectorizer_unique = TfidfVectorizer(min_df=1, stop_words='english', lowercase=True)
Xu_train_features = vectorizer_unique.fit_transform(Xu_train)
Xu_test_features = vectorizer_unique.transform(Xu_test)

model_unique = LogisticRegression(max_iter=1000)
model_unique.fit(Xu_train_features, Yu_train)
print('test accuracy on de-duplicated data:', accuracy_score(Yu_test, model_unique.predict(Xu_test_features)))
```
